### Verificar disponibilidad de ChromaDB

```python
# Verificar ChromaDB (check_chroma_simple)
import chromadb
from pathlib import Path

persist_dir = './data/chroma'
collection_name = 'normativa_laboral'

print('='*80)
print('VERIFICACIÓN DE DISPONIBILIDAD DE CHROMADB')
print('='*80)

if not Path(persist_dir).exists():
    print(f'❌ Directorio {persist_dir} NO EXISTE')
    db_available = False
else:
    try:
        client = chromadb.PersistentClient(path=persist_dir)
        collection = client.get_collection(name=collection_name)
        count = collection.count()
        print(f'✅ ChromaDB DISPONIBLE ({count} chunks)')
        db_available = count > 0
    except Exception as e:
        print(f'❌ Error: {e}')
        db_available = False

print('='*80)
```

## Diagnóstico: Tool Prestaciones Sociales (Debugging Detallado)

Test para diagnosticar por qué tool_results sale como null en LangSmith


```python
print('\n' + '='*80)
print('DIAGNÓSTICO DETALLADO: TOOL PRESTACIONES SOCIALES')
print('='*80)

# Importar módulos necesarios para debugging
from src.graph import classify_node, tool_calling_node
import re

# Usar la misma consulta
diagnostic_query = '¿Cómo calculo las prestaciones sociales con un salario de $2500000?'

print(f'\n📋 QUERY: {diagnostic_query}')

# PASO 1: Clasificación
print('\n' + '-'*80)
print('PASO 1: CLASIFICACIÓN')
print('-'*80)

state_step1 = {
    'query': diagnostic_query,
    'classification': '',
    'documents': [],
    'tool_results': None,
    'answer': '',
    'verification': {},
    'metadata': {}
}

state_step1 = classify_node(state_step1)
classification = state_step1.get('classification')
print(f'✓ Clasificación obtenida: {classification}')

# PASO 2: Tool Calling
print('\n' + '-'*80)
print('PASO 2: TOOL CALLING - DETECCIÓN DE HERRAMIENTAS')
print('-'*80)

state_step2 = {
    'query': diagnostic_query,
    'classification': classification,
    'documents': [],
    'tool_results': None,
    'answer': '',
    'verification': {},
    'metadata': {}
}

# Debugging manual de la detección
print(f'\n👀 Análisis manual de regex:')

# Palabras clave
keywords = ["prestaciones", "cesantías", "prima", "liquidación"]
has_keyword = any(word in diagnostic_query.lower() for word in keywords)
print(f'   Palabras clave encontradas: {has_keyword}')
for kw in keywords:
    if kw in diagnostic_query.lower():
        print(f'      ✓ "{kw}" encontrada')

# Patrones de salario
salary_patterns = [
    (r'\$\s*([\d,\.]+)', '$2500000'),
    (r'salario\s+(?:de\s+)?\$?([\d,\.]+)', 'salario de 2500000'),
    (r'([\d,\.]+)\s*(?:pesos|COP)', '2500000 pesos/COP'),
]

print(f'\n   Búsqueda de patrones de salario:')
for pattern, desc in salary_patterns:
    match = re.search(pattern, diagnostic_query, re.IGNORECASE)
    if match:
        print(f'      ✓ Patrón "{desc}" coincide: {match.group(0)}')
    else:
        print(f'      ✗ Patrón "{desc}" no coincide')

# Ejecutar tool_calling_node
print(f'\n→ Ejecutando tool_calling_node...')
state_step2 = tool_calling_node(state_step2)
tool_results = state_step2.get('tool_results')

print(f'\n✅ RESULTADO FINAL:')
print(f'   tool_results = {tool_results}')

if tool_results:
    print(f'   ✓ Tool detectada: {tool_results.get("tool_used")}')
    if tool_results.get("salario_detectado"):
        print(f'   ✓ Salario extraído: ${tool_results.get("salario_detectado"):,.2f}')
else:
    print(f'   ❌ tool_results es NULL - investigar por qué')

print('\n' + '='*80)
```
