## Sección 1: Inicializar la Base de Datos

### Verificar disponibilidad de ChromaDB

```python
# Verificar ChromaDB (check_chroma_simple)
import chromadb
from pathlib import Path

persist_dir = './data/chroma'
collection_name = 'normativa_laboral'

print('='*80)
print('VERIFICACIÓN DE DISPONIBILIDAD DE CHROMADB')
print('='*80)

if not Path(persist_dir).exists():
    print(f'❌ Directorio {persist_dir} NO EXISTE')
    db_available = False
else:
    try:
        client = chromadb.PersistentClient(path=persist_dir)
        collection = client.get_collection(name=collection_name)
        count = collection.count()
        print(f'✅ ChromaDB DISPONIBLE ({count} chunks)')
        db_available = count > 0
    except Exception as e:
        print(f'❌ Error: {e}')
        db_available = False

print('='*80)
```

### Inicializar Vectorstore

```python
#Construir el Grafo LangGraph
print('\n' + '='*80)
print('CONSTRUYENDO LANGGRAPH RAG')
print('='*80)

try:
    print('⏳ Compilando grafo...')
    graph = build_graph()
    print('✅ Grafo compilado exitosamente')
except Exception as e:
    print(f'❌ Error al construir grafo: {e}')
    
print('='*80)
```

### Caso 1: Búsqueda Básica

```python

print('\n' + '='*80)
print('CASO 1: BÚSQUEDA POR LEY ESPECÍFICA (search_by_document_type)')
print('='*80)

# Esta consulta debe activar search_by_document_type
query1 = 'Quiero saber sobre la ley 1010 de 2006'

try:
    print(f'\n💬 Consulta: {query1}')
    print('   Herramienta esperada: search_by_document_type')
    start = time.time()
    
    initial_state = {
        'query': query1,
        'classification': '',
        'documents': [],
        'tool_results': None,
        'answer': '',
        'verification': {},
        'metadata': {}
    }
    result = graph.invoke(initial_state)
    elapsed = time.time() - start
    
    print(f'\n📊 Herramientas ejecutadas: {result.get("tool_results")}\n')
    print(f'📊 Respuesta (clasificación: {result.get("classification", "N/A")}):\n')
    answer = result.get('answer', '')
    if len(answer) > 300:
        print(f'   {answer[:300]}...')
    else:
        print(f'   {answer}')
    print(f'\n⏱️ Tiempo: {elapsed:.3f}s')
    
except Exception as e:
    print(f'❌ Error: {e}')
    import traceback
    traceback.print_exc()
    
print('\n' + '='*80)
```
